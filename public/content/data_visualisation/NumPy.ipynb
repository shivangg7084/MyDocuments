## 3. Array attributes

For an array:
- `shape` → dimensions, e.g. `(2, 3)`
- `size` → total number of elements
- `ndim` → number of axes
- `dtype` → element data type
- `itemsize` → bytes occupied by one element

### Important distinction
A `(2, 3)` array has:
- `ndim = 2`
- `size = 6`


````python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("shape:", arr.shape)
print("size:", arr.size)
print("ndim:", arr.ndim)
print("dtype:", arr.dtype)
print("itemsize:", arr.itemsize)
````

## 4. Reshaping

`reshape()` changes the shape **without changing the number of elements**.

Rule:

`old_size == new_shape_product`

Example: 12 elements → `(3, 4)` because `3 × 4 = 12`.

You can use `-1` to let NumPy infer one dimension.


````python
arr = np.arange(1, 13)

print(arr.reshape(3, 4))
print(arr.reshape(2, 6))
print(arr.reshape(2, 2, 3))

# NumPy infers the second dimension
print(arr.reshape(3, -1))
````

## 5. Indexing

Python uses **0-based indexing**.

For 2D array:
```text
arr[row, column]
```

Useful patterns:
- `arr[2]` → third row / third element in 1D
- `arr[:, 0]` → first column
- `arr[1:, 2:]` → rows from index 1 and columns from index 2


````python
arr = np.arange(1, 13).reshape(3, 4)
print(arr)

print("first row:", arr[0])
print("first column:", arr[:, 0])
print("third column:", arr[:, 2])
print("last two columns:", arr[:, 2:])
print("bottom-right block:", arr[1:, 2:])
print("element:", arr[0, 3])
````

## 6. Slicing

General syntax:

```python
arr[start:stop:step]
```

`stop` is exclusive.

For 2D:
```python
arr[row_start:row_stop, col_start:col_stop]
```

### Reverse
```python
arr[::-1]
```

### Important
Basic NumPy slicing often produces a **view**, not an independent copy. Mutating the slice can mutate the original array.


````python
arr = np.arange(1, 13).reshape(3, 4)

print(arr[:, ::-1])       # reverse columns
print(arr[::-1, :])       # reverse rows
print(arr[::2, ::2])      # every second row/column
print(arr[0:2, 1:3])      # selected block
````

## 7. Boolean masking

Boolean indexing is one of NumPy's most important Data Science features.

```python
arr[arr > 10]
arr[arr % 2 == 0]
```

For multiple conditions use:
- `&` → AND
- `|` → OR
- `~` → NOT

**Always parenthesize each condition.**


````python
arr = np.arange(1, 11)

print(arr[arr > 5])
print(arr[arr % 2 == 0])
print(arr[(arr >= 3) & (arr <= 7)])
print(arr[(arr < 3) | (arr > 8)])
print(arr[~(arr % 2 == 0)])
````

## 8. Conditional replacement

### Direct masking
```python

```

### `np.where(condition, value_if_true, value_if_false)`

Unlike direct assignment, `where` conveniently creates a result array.


````python
arr = np.array([20, 40, 60, 80, 100])

arr2 = arr.copy()
arr2[arr2 > 50] = 0
print(arr2)

print(np.where(arr > 50, 100, arr))

idx = np.where(arr > 50)
print("indices:", idx)
````

## 11. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of compatible shapes.

Example:
```text
[1, 2, 3, 4] + 10
```
adds 10 to every element.

### 2D + 1D example
A `(2, 3)` matrix can be added to a `(3,)` vector because the last dimensions match.

### Broadcasting rule
Compare dimensions from the **right**:
Two dimensions are compatible if:
1. they are equal, or
2. one of them is `1`, or
3. one dimension is missing.

Otherwise NumPy raises a broadcasting error.


````python
arr = np.array([1, 2, 3, 4])
print(arr + 10)

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
vector = np.array([10, 20, 30])

print(matrix + vector)
````

## 12. Statistics

Core functions:
- `np.mean`
- `np.median`
- `np.std`
- `np.var`
- `np.sum`
- `np.min`
- `np.max`
- `np.argmax`
- `np.argmin`

### Standardization / Z-score

$$z = \frac{x - \mu}{\sigma}$$

After standardization, the values have approximately:
- mean = 0
- standard deviation = 1

### Min-Max normalization

$$x' = \frac{x-x_{min}}{x_{max}-x_{min}}$$

Result normally lies in `[0, 1]`.


````python
data = np.array([1, 2, 3, 4, 5])

mean = np.mean(data)
median = np.median(data)
std = np.std(data)
var = np.var(data)

z = (data - mean) / std
minmax = (data - data.min()) / (data.max() - data.min())

print(mean, median, std, var)
print("z-score:", z)
print("min-max:", minmax)
````

## 13. `axis=0` vs `axis=1`

For a 2D matrix:

```text
axis=0 → collapse rows → operate column-wise
axis=1 → collapse columns → operate row-wise
```

For a `(3, 4)` matrix:
- `sum(axis=0)` → shape `(4,)`
- `sum(axis=1)` → shape `(3,)`

**Memory trick:** axis tells NumPy which dimension to collapse.


````python
data = np.array([
    [80, 75, 90],
    [85, 88, 92],
    [78, 80, 85]
])

print("column sums:", np.sum(data, axis=0))
print("row sums:", np.sum(data, axis=1))
print("column means:", np.mean(data, axis=0))
print("row means:", np.mean(data, axis=1))
````

## 17. Copy vs View — very important

### Copy
```python
b = a.copy()
```
Independent memory. Changes to `b` do not affect `a`.

### View
```python
b = a.view()
```
Shares underlying data. Changes can affect `a`.

### Check memory sharing
```python
np.shares_memory(a, b)
```

This is a common interview question.


````python
arr = np.arange(1, 5)

copy_arr = arr.copy()
view_arr = arr.view()

copy_arr[0] = 100
view_arr[1] = 200

print("original:", arr)
print("copy:", copy_arr)
print("view:", view_arr)

print("copy shares memory:", np.shares_memory(arr, copy_arr))
print("view shares memory:", np.shares_memory(arr, view_arr))
````

## 20. Correlation matrix

`np.corrcoef(X)` computes correlation coefficients.

If observations are rows and features/subjects are columns, use:

```python
np.corrcoef(X, rowvar=False)
```

or equivalently:

```python
np.corrcoef(X.T)
```

Correlation ranges from `-1` to `+1`:
- `+1` → perfect positive linear relationship
- `0` → no linear correlation
- `-1` → perfect negative linear relationship

**Correlation does not imply causation.**


````python
data = np.array([
    [80, 75, 90],
    [85, 88, 92],
    [78, 80, 85],
    [90, 95, 98]
])

corr = np.corrcoef(data, rowvar=False)
print(corr)
````

# 21. Mini EDA with NumPy

Suppose each row is:

`[age, subject_1_marks, subject_2_marks]`

Typical EDA questions:
- extract a column
- filter rows
- calculate averages
- find maximum scorer
- replace invalid/low values
- count values satisfying a condition
- inspect correlations


````python
data = np.array([
    [22, 85, 90],
    [19, 78, 88],
    [25, 92, 95],
    [18, 70, 75],
    [21, 88, 91]
])

ages = data[:, 0]
print("ages:", ages)

print("younger than 21:\n", data[data[:, 0] < 21])

print("average of columns:", np.mean(data, axis=0))
print("average age:", np.mean(data[:, 0]))

print("third column > 90:\n", data[data[:, 2] > 90])

marks_only = data[:, 1:]
total_scores = np.sum(marks_only, axis=1)
best_idx = np.argmax(total_scores)

print("totals:", total_scores)
print("best student index:", best_idx)
print("best student row:", data[best_idx])

print("correlation:\n", np.corrcoef(marks_only, rowvar=False))
````

## 22. Important correction from the original practice

A common normalization mistake is operator precedence.

Incorrect:
```python
arr - np.min(arr) / np.max(arr) - np.min(arr)
```

Correct:
```python
(arr - arr.min()) / (arr.max() - arr.min())
```

Always use parentheses when implementing formulas.


````python
arr = np.array([10, 20, 30, 40, 50])

normalized = (arr - arr.min()) / (arr.max() - arr.min())
print(normalized)
````

````python
# HARD Q1 — Row-wise min-max normalization

X = np.array([
    [10, 20, 30],
    [5, 10, 15],
    [100, 150, 200]
], dtype=float)

row_min = X.min(axis=1, keepdims=True)
row_max = X.max(axis=1, keepdims=True)

row_norm = (X - row_min) / (row_max - row_min)
print(row_norm)
````

````python
# HARD Q5 — Column standardization

X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400]
], dtype=float)

mu = X.mean(axis=0, keepdims=True)
sigma = X.std(axis=0, keepdims=True)

Z = (X - mu) / sigma

print(Z)
print("means:", Z.mean(axis=0))
print("stds:", Z.std(axis=0))
````

````python
# HARD Q8 — Pairwise Euclidean distances

A = np.array([
    [0, 0],
    [1, 1],
    [2, 2]
], dtype=float)

B = np.array([
    [0, 1],
    [3, 3]
], dtype=float)

# A[:, None, :] -> (n, 1, d)
# B[None, :, :] -> (1, m, d)
# Result -> (n, m, d), then reduce over d
distances = np.sqrt(np.sum((A[:, None, :] - B[None, :, :]) ** 2, axis=2))

print(distances)
````

## Challenge E — Sliding-window maximum
Given a 1D array and window size `k`, calculate the maximum of every window.

First solve with a loop, then investigate a vectorized NumPy solution using `sliding_window_view`.


````python
# Expert Challenge A — solution

X = np.array([
    [10, 20, 30],
    [11, 21, 29],
    [10, 100, 31],
    [12, 22, 30]
], dtype=float)

z = (X - X.mean(axis=0)) / X.std(axis=0)

outlier_rows = X[np.any(np.abs(z) > 2, axis=1)]

print("Z-scores:\n", z)
print("outlier rows:\n", outlier_rows)
````

````python
# Expert Challenge B — nearest neighbor

X = np.array([
    [0, 0],
    [2, 2],
    [5, 5],
    [1, 1]
], dtype=float)

q = np.array([1.2, 1.1])

dist = np.linalg.norm(X - q, axis=1)
idx = np.argmin(dist)

print("nearest index:", idx)
print("nearest point:", X[idx])
print("distance:", dist[idx])
````

````python
# Expert Challenge C — feature engineering

X = np.array([
    [20, 50000, 1],
    [30, 80000, 6],
    [40, 100000, 12]
], dtype=float)

age = X[:, 0]
income = X[:, 1]
experience = X[:, 2]

exp_age = (experience / age)[:, None]
income_z = ((income - income.mean()) / income.std())[:, None]

features = np.column_stack((X, exp_age, income_z))

print(features)
````

````python
# Expert Challenge D — group means

groups = np.array([0, 1, 0, 1, 0])
scores = np.array([80, 90, 70, 100, 85])

for g in np.unique(groups):
    print(g, scores[groups == g].mean())
````

````python
# Expert Challenge E — sliding-window maximum
# NumPy >= 1.20

from numpy.lib.stride_tricks import sliding_window_view

arr = np.array([2, 1, 5, 3, 4, 6, 1])
k = 3

windows = sliding_window_view(arr, k)
result = windows.max(axis=1)

print(windows)
print(result)
````

# 26. Common mistakes

### Mistake 1 — `reshape()` without assignment
```python
arr.reshape(2, 3)
```
may return a reshaped array without changing `arr` itself.

Use:
```python
arr = arr.reshape(2, 3)
```

### Mistake 2 — Wrong boolean operators
Wrong:
```python
arr > 3 and arr < 8
```

Correct:
```python
(arr > 3) & (arr < 8)
```

### Mistake 3 — Forgetting parentheses
Wrong:
```python
arr > 3 & arr < 8
```

Correct:
```python
(arr > 3) & (arr < 8)
```

### Mistake 4 — Confusing `axis`
For a table:
- `axis=0` → each column
- `axis=1` → each row

### Mistake 5 — Assuming slices are always copies
NumPy views can share memory.

### Mistake 6 — Using Python loops unnecessarily
Prefer vectorization, broadcasting, reductions and boolean indexing.

### Mistake 7 — Division by zero in normalization
If `max == min`, min-max normalization has a zero denominator. Handle constant columns explicitly.

### Mistake 8 — Confusing `argmax` with `max`
- `max` → value
- `argmax` → index
