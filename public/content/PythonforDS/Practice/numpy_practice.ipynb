```python
data = np.array([
    [80, 75, 90],
    [85, 88, 92],
    [78, 80, 85],
    [90, 95, 98]
]) 
```

```python
# Find the average marks in each subject: 
avg_arr = np.mean(data,axis = 0)
print(avg_arr)
```

```python
#  Find the average marks of each student.
student_avg = np.mean(data,axis=1)
print(student_avg)
```

```python
# Find the student with the highest total marks.
total_marks = np.sum(data,axis=1)
print(total_marks)
highest = np.argmax(total_marks)
print(highest+1)
```

```python
# Count how many students scored more than 85 in the third subject. 
count = np.sum(data[:,2]>85)
print(count)
```

```python
# Find the maximum marks in each subject. 
maximum = np.max(data,axis=0)
print(maximum)
```

```python
# Find the minimum marks in each subject.
minimum = np.min(data,axis=0)
print(minimum)
```

```python
# Find students whose average marks are greater than 85.
avg_marks = np.mean(data,axis = 1)
students = data[avg_marks>85]
print(students)
```

```python
# Replace all marks below 80 with 0.
new_data = data.copy()
new_data[new_data<80] = 0
print(new_data)
```

```python
# Count how many values are greater than the overall mean.
overall_mean = np.mean(data)
count = np.sum(data > overall_mean)
print(count)
```

```python
# Find the correlation matrix of the subjects. 
correlation = np.corrcoef(data.T)
print(correlation)
```

```python
 # Day 4 (Mini EDA)
data = np.array([
    [22, 85, 90],
    [19, 78, 88],
    [25, 92, 95],
    [18, 70, 75],
    [21, 88, 91]
]) 
```

```python
# Extract the age column.
print(data[:,0])
```

```python
# Find students younger than 21. 
arr = data[data[:,0]<21]
print(arr)
```

```python
# Find the average marks in each subject. 
data = np.array([
    [22, 85, 90],
    [19, 78, 88],
    [25, 92, 95],
    [18, 70, 75],
    [21, 88, 91]
]) 

avg_marks = np.mean(data, axis=0)

print(avg_marks)
```

```python
# Find the average age. 
avg_age = np.mean(data[:,0])
print(avg_age)
```

```python
# Find students who scored more than 90 in the third column. 
print(data[data[:,2]>90])
```

```python
# Replace all marks below 80 with 0. 
data[data<80] = 0
print(data)
```

```python
# Find the student with the highest total score. 
total_scores = np.sum(data[:,1:],axis=1)
print(total_scores)
highest_student = np.argmax(total_scores)
print(highest_student)
print(data[highest_student])
```

```python
# Normalize the second column. 
new_data = data[:,2]
print(new_data)
normalized_arr = new_data - np.min(new_data)/np.max(new_data)-np.min(new_data)
print(normalized_arr)
```

```python
# Count how many values are greater than the overall mean. 
overall_mean = np.mean(data)
count = np.sum(data > overall_mean)
print(overall_mean)
print(count)
```

```python
# Find the correlation matrix 
new_arr = data.T 
print(np.corrcoef(new_arr))
```
