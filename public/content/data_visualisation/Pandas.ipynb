Numeric: mean, median, model-based or domain-specific.

# 20. GroupBy — Split, Apply, Combine

`groupby()` is a core Pandas concept.

```text
Split → Apply → Combine
```

Examples:
- average salary by department
- sales by region
- count users by category


````python
df = pd.DataFrame({
    "Product":["A","A","B","B","B"],
    "Region":["North","South","North","North","South"],
    "Value":[10,20,30,40,50]
})
print(df.groupby("Product")["Value"].mean())
print(df.groupby(["Product","Region"])["Value"].mean())
````

## Multiple aggregations and named aggregation

````python
print(df.groupby("Region")["Value"].agg(["mean","sum","count","min","max"]))

result = df.groupby("Product").agg(
    mean_value=("Value","mean"),
    total_value=("Value","sum"),
    n=("Value","count")
)
print(result)
````

# 21. `transform()` — Very Important for ML

Unlike aggregation, `transform()` returns a Series with the **same length/index as the original data**.

This makes it ideal for group-wise:
- normalization
- imputation
- statistics
- target-related features when leakage is controlled


````python
df = pd.DataFrame({
    "department":["IT","IT","HR","HR"],
    "salary":[60000,80000,50000,70000]
})
df["dept_mean"] = df.groupby("department")["salary"].transform("mean")
df["salary_minus_dept_mean"] = df["salary"] - df["dept_mean"]
print(df)
````

# 22. Merge / Join

`merge()` is conceptually similar to SQL JOIN.

- `inner` → matching keys
- `left` → all left keys
- `right` → all right keys
- `outer` → union of keys

### Critical ML/data-engineering issue
Many-to-many joins can unexpectedly multiply rows.


````python
df1 = pd.DataFrame({"Key":["A","B","C"],"Value1":[1,2,3]})
df2 = pd.DataFrame({"Key":["A","B","D"],"Value2":[4,5,6]})

print(pd.merge(df1,df2,on="Key",how="inner"))
print(pd.merge(df1,df2,on="Key",how="left"))
print(pd.merge(df1,df2,on="Key",how="right"))
print(pd.merge(df1,df2,on="Key",how="outer"))
````

## Merge validation

Use `validate=` to catch incorrect relationship assumptions:

```python
validate="one_to_one"
validate="one_to_many"
validate="many_to_one"
validate="many_to_many"
```

This is an excellent production-data habit.


````python
customers = pd.DataFrame({"customer_id":[1,2,3],"name":["A","B","C"]})
orders = pd.DataFrame({"customer_id":[1,1,2],"amount":[100,200,150]})

merged = customers.merge(
    orders, on="customer_id", how="left",
    validate="one_to_many"
)
print(merged)
````

# 29. Outlier Detection — IQR

`IQR = Q3 - Q1`

Lower bound:
`Q1 - 1.5 * IQR`

Upper bound:
`Q3 + 1.5 * IQR`

Values outside are **potential** outliers, not automatically invalid records.


````python
s = pd.Series([10,12,13,14,15,16,100])
q1, q3 = s.quantile([0.25,0.75])
iqr = q3-q1
lower = q1-1.5*iqr
upper = q3+1.5*iqr
print("outliers:", s[(s<lower)|(s>upper)].tolist())
````

# 37. ML Data Leakage

Leakage occurs when information unavailable at prediction time enters model training.

Common causes:
1. scaling/imputation using full data
2. target encoding using full data
3. future information in time-series features
4. duplicate users appearing across train/test
5. target accidentally included in features

### Safe conceptual workflow

`Split → fit preprocessing on train → transform train/test → train model`

Use sklearn `Pipeline` and `ColumnTransformer` for robust production workflows.


````python
from sklearn.model_selection import train_test_split

df = pd.DataFrame({
    "age":[20,21,22,23,24,25,26,27],
    "income":[20,22,24,np.nan,30,32,34,36],
    "target":[0,0,0,1,1,1,1,1]
})

train, test = train_test_split(df, test_size=0.25, random_state=42)

median = train["income"].median()
train["income"] = train["income"].fillna(median)
test["income"] = test["income"].fillna(median)

print(train)
print(test)
````

# 38. Pandas + sklearn preprocessing

Typical production architecture:

```text
DataFrame
   ↓
Train/Test Split
   ↓
ColumnTransformer
 ┌─────────────┬──────────────┐
 │ Numeric     │ Categorical  │
 │ Imputation  │ Imputation   │
 │ Scaling     │ One-Hot      │
 └─────────────┴──────────────┘
   ↓
Model
```

Pandas is excellent for exploration and feature creation. sklearn pipelines are usually preferable for reproducible model preprocessing.


````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.DataFrame({
    "age":[20,25,30,35,40,45],
    "income":[30000,40000,np.nan,60000,70000,80000],
    "city":["Delhi","Mumbai","Delhi","Pune","Mumbai","Delhi"],
    "target":[0,0,1,1,1,1]
})

X = df.drop(columns="target")
y = df["target"]

numeric = ["age","income"]
categorical = ["city"]

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical)
])

X_transformed = preprocessor.fit_transform(X)
print(X_transformed.shape)
````

# 41. Data Validation Checklist

Before training:

### Structure
- expected rows/columns?
- correct column names?

### Types
- numeric actually numeric?
- dates datetime?

### Missingness
- where and how much?
- is missingness meaningful?

### Duplicates
- valid repeated events or accidental duplicates?

### Ranges
- impossible ages?
- impossible percentages?
- negative prices?

### Categories
- unexpected categories?
- rare categories?

### Target
- imbalance?
- leakage?

````python
# Expert B — RFM
transactions = pd.DataFrame({
    "customer_id":[1,1,1,2,2,3],
    "date":pd.to_datetime([
        "2026-08-01","2026-08-10","2026-08-20",
        "2026-08-05","2026-08-15","2026-08-18"
    ]),
    "amount":[100,150,200,80,120,500]
})

reference_date = transactions["date"].max() + pd.Timedelta(days=1)

rfm = transactions.groupby("customer_id").agg(
    last_purchase=("date","max"),
    frequency=("date","count"),
    monetary=("amount","sum")
)
rfm["recency"] = (reference_date - rfm["last_purchase"]).dt.days
print(rfm.drop(columns="last_purchase"))
````

````python
# Expert C — historical features
t = pd.DataFrame({
    "customer":["A","A","A","B","B"],
    "date":pd.to_datetime([
        "2026-01-01","2026-01-03","2026-01-06",
        "2026-01-01","2026-01-04"
    ]),
    "amount":[100,200,150,300,400]
}).sort_values(["customer","date"])

g = t.groupby("customer", group_keys=False)
t["previous"] = g["amount"].shift(1)
t["previous_count"] = g.cumcount()
t["rolling_7_sum"] = (
    t.set_index("date")
     .groupby("customer")["amount"]
     .rolling("7D")
     .sum()
     .reset_index(level=0, drop=True)
     .reset_index(drop=True)
)
print(t)
````

````python
# Expert D — rare categories
df = pd.DataFrame({"category":["A"]*80 + ["B"]*15 + ["C"]*4 + ["Rare"]})
freq = df["category"].value_counts(normalize=True)
rare = freq[freq < 0.01].index
df["category_clean"] = df["category"].where(
    ~df["category"].isin(rare), "Other"
)
print(df["category_clean"].value_counts())
````

````python
# Expert G — latest ingestion record
events = pd.DataFrame({
    "customer_id":[1,1,1,2],
    "timestamp":pd.to_datetime([
        "2026-01-01 10:00","2026-01-01 10:00",
        "2026-01-02 10:00","2026-01-01 12:00"
    ]),
    "event_type":["login","login","purchase","login"],
    "ingestion_time":pd.to_datetime([
        "2026-01-01 10:05","2026-01-01 10:10",
        "2026-01-02 10:05","2026-01-01 12:05"
    ])
})
events = events.sort_values("ingestion_time")
deduped = events.drop_duplicates(
    ["customer_id","timestamp","event_type"], keep="last"
)
print(deduped)
````
