| Operation              | Time Complexity                           |
| ---------------------- | ----------------------------------------- |
| Access by index        | O(1)                                      |
| Update by index        | O(1)                                      |
| Append                 | O(1) (amortized)                          |
| Pop from end           | O(1)                                      |
| Insert at beginning    | O(n)                                      |
| Remove by value        | O(n)                                      |
| Search (`in`, `index`) | O(n)                                      |
| Sort                   | O(n log n)                                |
| Reverse                | O(n)                                      |
| Copy                   | O(n)                                      |
| Slice                  | O(k) (where *k* is the size of the slice) |


| Concept            | Example            |
| ------------------ | ------------------ |
| Create             | `a = [1, 2, 3]`    |
| Index              | `a[0]`             |
| Negative Index     | `a[-1]`            |
| Slice              | `a[1:4]`           |
| Reverse Slice      | `a[::-1]`          |
| Append             | `a.append(5)`      |
| Extend             | `a.extend([6, 7])` |
| Insert             | `a.insert(1, 100)` |
| Remove             | `a.remove(2)`      |
| Pop                | `a.pop()`          |
| Clear              | `a.clear()`        |
| Sort               | `a.sort()`         |
| Reverse            | `a.reverse()`      |
| Copy               | `b = a.copy()`     |
| Membership         | `3 in a`           |
| Length             | `len(a)`           |
| Sum                | `sum(a)`           |
| Max                | `max(a)`           |
| Min                | `min(a)`           |
| Iteration          | `for x in a:`      |
| List Comprehension | `[x*x for x in a]` |
