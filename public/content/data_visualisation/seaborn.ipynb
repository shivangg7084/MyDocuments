# Seaborn

- Seaborn is a Python visualisation library based on matplotlib that provides a high-level interface for drawing attractive and informative statiscal graphics. 
- Seaborn helps in creating complex visualisations with just a line of code.

!pip install seaborn

```python
import seaborn as sns
import pandas as pd
```

```python
### Basic ploting 
# tips = sns.load_dataset('tips')
df = pd.read_csv('data.csv')
df.head()
```

```python
# scatter plot 

import matplotlib.pyplot as plt 

sns.scatterplot(x='Value',y='Sales',data=df)
plt.title("Scatter Plot of Value vs Sales")
plt.show()
```

```python
# Line plot 
sns.lineplot(x='Product',y='Sales',data=df)
plt.title("Line Plot of Product vs Sales")
plt.show()
```

```python
## Categorical variable
```

```python
sns.barplot(x='Region',y='Sales',data=df)
plt.title("Total Sales as per the Region")
plt.show()
```

```python
#Box plot -> Outliers
```

```python
sns.boxplot(x='Region',y='Sales',data=df)
```

```python
# Violin plot -> distribution of data
```

```python
sns.violinplot(x='Region',y='Sales',data=df)
```

```python
## Histogram
```

```python
sns.histplot(df['Sales'],bins=3,kde=True)
```

```python
## KDE plot
```

```python
sns.kdeplot(df['Sales'],fill=True)
```

```python
## Pairplot
```

```python
sns.pairplot(df)
```

```python
## Heatmap 
corr = df[['Sales','Value']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Product',y='Sales',data=df,estimator=sum)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Region',y='Sales',data=df,estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```
